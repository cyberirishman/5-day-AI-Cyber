## Step 1 — The program

Run the cell below. It will **stop and wait** for you: a small box appears asking for a number.
Type one, press **Enter**, and it will ask for the second.

> **This surprises everyone the first time.** While a cell is waiting for you to type, it looks
> like it has frozen — the ▶ button turns into a spinning circle. It has not crashed. It is
> waiting. If you ever want to stop a waiting cell, press the ⏹ stop button.


````python
# ---- First, tell the person what this program is going to do -----------------
# Never leave a user guessing what a program wants from them. Say it up front.
# print() just displays a line of text underneath the cell.
print("This program adds two numbers together.")
print("I will ask you for two numbers, one at a time, and then show you the total.")
print()                       # an empty print() leaves a blank line, for readability

# ---- Ask the user for the first number ---------------------------------------
# input() shows a box, waits for the person to type something, and hands back
# whatever they typed. The text in the brackets is the prompt they see.
#
# IMPORTANT: input() ALWAYS gives back TEXT, even when the person types a number.
# float() converts that text into a number we can do maths with.
#   float("7")    ->  7.0     (a number)
#   float("2.5")  ->  2.5     (decimals are fine)
# We explain WHY this matters in Step 2 -- it is the classic first bug.
first_number = float(input("Type your first number, then press Enter: "))

# ---- Ask for the second number, exactly the same way -------------------------
second_number = float(input("Now type your second number, then press Enter: "))

# ---- Do the maths ------------------------------------------------------------
# The '=' sign does NOT mean "equals" here. It means "work out the right-hand
# side, then store the result under the name on the left".
total = first_number + second_number

# ---- Show the answer ---------------------------------------------------------
# print() displays something underneath the cell.
# An f-string (the f before the quote) lets us drop values straight into text:
# anything inside {curly braces} is replaced by its value.
print(f"{first_number} + {second_number} = {total}")
````

## Step 2 — Why `float()`? The bug that catches everyone

Look again at that line:

```python
first_number = float(input("..."))
```

Without `float()`, Python would treat what you typed as **text**, not a number. And Python's `+`
does something different to text than it does to numbers:

- with **numbers**, `+` adds → `2 + 3` is `5`
- with **text**, `+` glues → `"2" + "3"` is `"23"`

So a program that forgets to convert does not crash. **It quietly gives you the wrong answer** —
which is far worse, because nothing tells you something went wrong.

Run the cell below to see both behaviours side by side. It needs no typing from you.


````python
# Two pieces of TEXT. The quote marks are what make them text rather than numbers.
text_two  = "2"
text_three = "3"

# Gluing text together with + :
print("Text  :", text_two, "+", text_three, "=", text_two + text_three, "  <-- glued, not added!")

# The same two values converted into numbers first:
number_two   = float(text_two)
number_three = float(text_three)
print("Number:", number_two, "+", number_three, "=", number_two + number_three, "  <-- actually added")

# type() tells you what Python thinks something is. This is a genuinely useful
# habit when a program misbehaves: ask Python what it is holding.
print()
print("Python thinks", repr(text_two),   "is a", type(text_two).__name__,  "(str = text)")
print("Python thinks", repr(number_two), "is a", type(number_two).__name__, "(float = number)")
````

## Step 3 — What if someone types "banana"?

`float("banana")` cannot work — there is no number in there — so Python stops with a `ValueError`
and the rest of the cell never runs.

Real programs expect people to type the wrong thing. The usual fix is **try / except**: *try*
something, and if it fails, do something sensible instead of falling over.

Run the cell below and deliberately type something silly the first time. It will ask again.


````python
# A function is a named, reusable block of code. We define it once here...
def ask_for_a_number(prompt):
    """Keep asking until the person actually types a number."""

    # while True: means "repeat forever" -- until a return statement escapes it.
    while True:
        answer = input(prompt)          # get whatever they typed (as text)

        try:
            # TRY to convert it. If this works, return the number and we are done:
            # return exits the function immediately, which also ends the loop.
            return float(answer)

        except ValueError:
            # ...and if float() could not do it, we land here instead of crashing.
            print(f"  '{answer}' is not a number. Try again -- digits only, e.g. 12 or 3.5")


# ...and now we can use it twice, which saves writing the same logic out again.
# Same courtesy as before: say what is about to happen before asking for anything.
print("This program adds two numbers together.")
print("I will ask you for two numbers, one at a time, and then show you the total.")
print()

a = ask_for_a_number("First number:  ")
b = ask_for_a_number("Second number: ")

print(f"\n{a} + {b} = {a + b}")
````
